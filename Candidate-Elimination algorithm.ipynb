```python
import csv

def is_consistent(h, instance):
    return all(h[i] == '?' or h[i] == instance[i] for i in range(len(h)))

def is_more_general_or_equal(g, s):
    return all(g[i] == '?' or g[i] == s[i] for i in range(len(g)))

def remove_more_general(hypotheses):
    unique_h = []
    for h in hypotheses:
        if h not in unique_h:
            unique_h.append(h)

    filtered_G = []
    for h1 in unique_h:
        if not any(h1 != h2 and is_more_general_or_equal(h1, h2) for h2 in unique_h):
            filtered_G.append(h1)
    return filtered_G

def learn(concepts, target):
    num_attributes = len(concepts[0])
    specific_h = ['0'] * num_attributes
    general_h = [["?" for _ in range(num_attributes)]]

    print("Initial Specific Boundary (S0):", specific_h)
    print("Initial General Boundary  (G0):", general_h)
    print("-" * 50)

    for i, h in enumerate(concepts):
        if target[i].strip().lower() == "yes":
            general_h = [g for g in general_h if is_consistent(g, h)]
            for x in range(num_attributes):
                if specific_h[x] == '0':
                    specific_h[x] = h[x]
                elif h[x] != specific_h[x]:
                    specific_h[x] = '?'

        elif target[i].strip().lower() == "no":
            temp_G = []
            for g in general_h:
                if is_consistent(g, h):
                    for x in range(num_attributes):
                        if g[x] == '?':
                            if specific_h[x] != '?' and specific_h[x] != '0' and specific_h[x] != h[x]:
                                new_g = g.copy()
                                new_g[x] = specific_h[x]
                                temp_G.append(new_g)
                            elif specific_h[x] == '?':
                                possible_values = list(set(c[x] for c in concepts))
                                for val in possible_values:
                                    if val != h[x]:
                                        new_g = g.copy()
                                        new_g[x] = val
                                        temp_G.append(new_g)
                else:
                    temp_G.append(g)

            general_h = remove_more_general(temp_G)

        print(f"\nStep {i+1} (Example: {h} -> {target[i]}):")
        print(f"S_{i+1} =", specific_h)
        print(f"G_{i+1} =", general_h)

    final_G = [g for g in general_h if is_more_general_or_equal(g, specific_h)]
    return specific_h, final_G

# --- Main Setup ---
csv_filename = "data.csv"

# Write dataset file
with open(csv_filename, 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Size", "Color", "Shape", "Class / Label"])
    writer.writerow(["Big", "Red", "Circle", "No"])
    writer.writerow(["Small", "Red", "Triangle", "No"])
    writer.writerow(["Small", "Red", "Circle", "Yes"])
    writer.writerow(["Big", "Blue", "Circle", "No"])
    writer.writerow(["Small", "Blue", "Circle", "Yes"])

# Read dataset file
concepts = []
target = []
with open(csv_filename, 'r') as file:
    reader = csv.reader(file)
    header = next(reader)
    for row in reader:
        concepts.append(row[:-1])
        target.append(row[-1])

# Run candidate elimination
s_final, g_final = learn(concepts, target)

print("\n" + "="*50)
print("FINAL SPECIFIC BOUNDARY (S):", s_final)
print("FINAL GENERAL BOUNDARY  (G):", g_final)
```
